```python
def generate_inventory(total_records):

    inventory_data = []

    for inventory_id in range(1, total_records + 1):

        # Progress
        if inventory_id % 10000 == 0:
            print(f"Generated {inventory_id:,} records...")

        # Random Dimension Records
        product = random.choice(product_list)
        warehouse = random.choice(warehouse_list)
        employee = random.choice(employee_list)
        date = random.choice(date_list)

        # Product Information
        supplier_id = product["supplier_id"]
        unit_cost = product["unit_cost"]

        # Transaction Type
        transaction_type = random.choices(
            population=list(transaction_types.keys()),
            weights=list(transaction_types.values()),
            k=1
        )[0]

        # Opening Stock
        opening_stock = random.randint(1000,10000)

        # Quantity In / Out
        quantity_in = 0
        quantity_out = 0

        if transaction_type in ["Purchase","Return"]:

            quantity_in = random.randint(50,1000)

        elif transaction_type in ["Sales","Production","Transfer"]:

            quantity_out = random.randint(50,1000)

        else:

            if random.random() < 0.5:
                quantity_in = random.randint(10,500)
            else:
                quantity_out = random.randint(10,500)

        # Closing Stock
        closing_stock = opening_stock + quantity_in - quantity_out

        if closing_stock < 0:
            closing_stock = 0

        # Inventory Value
        inventory_value = round(
            closing_stock * unit_cost,
            2
        )

        # Reorder Level
        reorder_level = random.randint(500,1500)

        # Stock Status
        if closing_stock < reorder_level:

            stock = "Low Stock"

        elif closing_stock <= 5000:

            stock = "Normal"

        else:

            stock = "Over Stock"

        # Batch Number
        batch_number = "BAT" + str(random.randint(100000,999999))

        # Expiry Days
        expiry_days = random.randint(30,730)

        # Remarks
        remarks = random.choice(remarks_list)

        # Append Record
        inventory_data.append({

            "inventory_id": inventory_id,

            "transaction_date_id": date["date_id"],

            "product_id": product["product_id"],

            "warehouse_id": warehouse["warehouse_id"],

            "supplier_id": supplier_id,

            "transaction_type": transaction_type,

            "opening_stock": opening_stock,

            "quantity_in": quantity_in,

            "quantity_out": quantity_out,

            "closing_stock": closing_stock,

            "unit_cost": unit_cost,

            "inventory_value": inventory_value,

            "reorder_level": reorder_level,

            "stock_status": stock,

            "batch_number": batch_number,

            "expiry_days": expiry_days,

            "employee_id": employee["employee_id"],

            "remarks": remarks

        })

    df = pd.DataFrame(inventory_data)

    return df
```
