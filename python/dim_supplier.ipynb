```python
import pandas as pd
import numpy as np
from faker import Faker
import random
```

```python
# =====================================================
# Manufacturing Production Intelligence Platform
# Table Name : dim_supplier
# Author : Mahesh Yangundi
# =====================================================
```

```python
# Initialize Faker
fake = Faker("en_IN")

# Number of Suppliers
NUM_SUPPLIERS = 250

# Lists
supplier_material = [
    "Cotton Yarn",
    "Polyester Yarn",
    "Reactive Dyes",
    "Chemicals",
    "Packaging Material",
    "Labels",
    "Fabric",
    "Threads",
    "Elastic",
    "Cartons"
]

cities = [
    "Solapur","Mumbai","Pune","Ichalkaranji","Nagpur",
    "Ahmedabad","Surat","Coimbatore","Erode","Tiruppur",
    "Panipat","Ludhiana","Jaipur","Indore","Hyderabad"
]

states = [
    "Maharashtra","Maharashtra","Maharashtra","Maharashtra",
    "Maharashtra","Gujarat","Gujarat","Tamil Nadu",
    "Tamil Nadu","Tamil Nadu","Haryana","Punjab",
    "Rajasthan","Madhya Pradesh","Telangana"
]

status = ["Active","Inactive"]

supplier_list = []

for i in range(1, NUM_SUPPLIERS + 1):

    city_index = random.randint(0, len(cities)-1)

    supplier = {

        "supplier_id": i,

        "supplier_code": f"SUP{i:04d}",

        "supplier_name": fake.company(),

        "material_supplied": random.choice(supplier_material),

        "city": cities[city_index],

        "state": states[city_index],

        "country": "India",

        "supplier_rating": round(random.uniform(3.50,5.00),2),

        "lead_time_days": random.randint(2,30),

        "phone": fake.phone_number(),

        "email": f"supplier{i}@gmail.com",

        "supplier_status": random.choice(status)

    }

    supplier_list.append(supplier)

# Create DataFrame
df = pd.DataFrame(supplier_list)

print("Supplier Table Generated Successfully!")
```

```python
print("Shape :", df.shape)

print("\nMissing Values")
print(df.isnull().sum())

print("\nSupplier Status")
print(df["supplier_status"].value_counts())

print("\nAverage Rating")
print(df["supplier_rating"].mean())

print("\nLead Time")
print(df["lead_time_days"].describe())
```

```python
df.to_csv("dim_supplier.csv", index=False)

df.to_excel("dim_supplier.xlsx", index=False)

print("dim_supplier.csv created successfully!")
print("dim_supplier.xlsx created successfully!")
```
