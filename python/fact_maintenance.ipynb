```python
def generate_maintenance(total_records):

    maintenance_data = []

    for maintenance_id in range(1, total_records + 1):

        # Progress
        if maintenance_id % 10000 == 0:
            print(f"Generated {maintenance_id:,} records...")

        # Random Dimension Records
        machine = random.choice(machine_list)

        employee = random.choice(employee_list)

        date = random.choice(date_list)

        # Maintenance Type
        maintenance_type = random.choices(
            population=list(maintenance_types.keys()),
            weights=list(maintenance_types.values()),
            k=1
        )[0]

        # Maintenance Status
        status = random.choices(
            population=list(maintenance_status.keys()),
            weights=list(maintenance_status.values()),
            k=1
        )[0]

        # Breakdown Hours
        breakdown_hours = round(
            random.uniform(0,24),
            2
        )

        # Repair Hours
        repair_hours = round(
            random.uniform(1,12),
            2
        )

        # Costs
        labor_cost = round(
            random.uniform(500,10000),
            2
        )

        spare_parts_cost = round(
            random.uniform(0,50000),
            2
        )

        total_cost = round(
            labor_cost + spare_parts_cost,
            2
        )

        # Priority
        priority = random.choices(
            population=list(priority_levels.keys()),
            weights=list(priority_levels.values()),
            k=1
        )[0]

        # Machine Condition
        machine_condition = random.choice(
            machine_conditions
        )

        # Next Service
        next_service_days = random.randint(
            30,
            180
        )

        # Result
        maintenance_result = random.choice(
            maintenance_results
        )

        # Parts
        parts_replaced = random.choice(
            parts_list
        )

        # Remarks
        remarks = random.choice(
            remarks_list
        )

        # Vendor
        vendor_required = random.choice(
            ["Yes","No"]
        )

        maintenance_data.append({

            "maintenance_id": maintenance_id,

            "maintenance_date_id": date["date_id"],

            "machine_id": machine["machine_id"],

            "employee_id": employee["employee_id"],

            "maintenance_type": maintenance_type,

            "maintenance_status": status,

            "breakdown_hours": breakdown_hours,

            "repair_hours": repair_hours,

            "labor_cost": labor_cost,

            "spare_parts_cost": spare_parts_cost,

            "total_cost": total_cost,

            "priority": priority,

            "machine_condition": machine_condition,

            "next_service_days": next_service_days,

            "maintenance_result": maintenance_result,

            "parts_replaced": parts_replaced,

            "remarks": remarks,

            "vendor_required": vendor_required

        })

    df = pd.DataFrame(maintenance_data)

    return df
```
