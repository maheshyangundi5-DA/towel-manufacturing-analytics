```python
import pandas as pd
from faker import Faker
import random

fake = Faker("en_IN")
```

```python
# =====================================================
# Manufacturing Production Intelligence Platform
# Table : dim_warehouse
# =====================================================
```

```python
warehouse_data = [

    ("WH001","Raw Material Warehouse 1","Raw Material","Solapur","Maharashtra"),
    ("WH002","Raw Material Warehouse 2","Raw Material","Solapur","Maharashtra"),
    ("WH003","Raw Material Warehouse 3","Raw Material","Pune","Maharashtra"),
    ("WH004","Raw Material Warehouse 4","Raw Material","Mumbai","Maharashtra"),

    ("WH005","Finished Goods Warehouse 1","Finished Goods","Solapur","Maharashtra"),
    ("WH006","Finished Goods Warehouse 2","Finished Goods","Pune","Maharashtra"),
    ("WH007","Finished Goods Warehouse 3","Finished Goods","Hyderabad","Telangana"),
    ("WH008","Finished Goods Warehouse 4","Finished Goods","Surat","Gujarat"),
    ("WH009","Finished Goods Warehouse 5","Finished Goods","Delhi","Delhi"),

    ("WH010","Packaging Warehouse 1","Packaging","Solapur","Maharashtra"),
    ("WH011","Packaging Warehouse 2","Packaging","Ahmedabad","Gujarat"),

    ("WH012","Export Warehouse 1","Export","Mumbai","Maharashtra"),
    ("WH013","Export Warehouse 2","Export","Chennai","Tamil Nadu"),

    ("WH014","Spare Parts Warehouse 1","Spare Parts","Solapur","Maharashtra"),
    ("WH015","Spare Parts Warehouse 2","Spare Parts","Nagpur","Maharashtra")
]

warehouse_list = []

for i, wh in enumerate(warehouse_data, start=1):

    warehouse_list.append({

        "warehouse_id": i,

        "warehouse_code": wh[0],

        "warehouse_name": wh[1],

        "warehouse_type": wh[2],

        "city": wh[3],

        "state": wh[4],

        "country": "India",

        "capacity_sqft": random.randint(5000,50000),

        "manager_name": fake.name(),

        "phone": fake.phone_number(),

        "warehouse_status": random.choice(["Active","Inactive"])

    })

df = pd.DataFrame(warehouse_list)

print("Warehouse Table Generated Successfully!")
```

```python
df.head()
```

```python
print("Shape :", df.shape)

print("\nMissing Values")
print(df.isnull().sum())

print("\nWarehouse Types")
print(df["warehouse_type"].value_counts())

print("\nWarehouse Status")
print(df["warehouse_status"].value_counts())
```

```python
df.to_csv("dim_warehouse.csv", index=False)

df.to_excel("dim_warehouse.xlsx", index=False)

print("dim_warehouse.csv created successfully!")
print("dim_warehouse.xlsx created successfully!")
```
