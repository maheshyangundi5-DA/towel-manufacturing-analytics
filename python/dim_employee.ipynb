```python
import pandas as pd
from faker import Faker
import random
```

```python
# =====================================================
# Manufacturing Production Intelligence Platform
# Table : dim_employee
# =====================================================
```

```python
fake = Faker("en_IN")

NUM_EMPLOYEES = 600

departments = (
    ["Production"] * 250 +
    ["Sales"] * 100 +
    ["Quality"] * 80 +
    ["Maintenance"] * 70 +
    ["Warehouse"] * 50 +
    ["HR"] * 20 +
    ["Finance"] * 20 +
    ["Management"] * 10
)

random.shuffle(departments)

designations = {
    "Production": ["Operator", "Senior Operator", "Production Supervisor"],
    "Sales": ["Sales Executive", "Sales Officer", "Sales Manager"],
    "Quality": ["Quality Inspector", "QA Engineer"],
    "Maintenance": ["Technician", "Maintenance Engineer"],
    "Warehouse": ["Store Keeper", "Warehouse Executive"],
    "HR": ["HR Executive", "HR Manager"],
    "Finance": ["Accountant", "Finance Executive"],
    "Management": ["Plant Manager", "General Manager"]
}

employee_status = ["Active", "Inactive", "Resigned"]

shifts = ["Morning", "Evening", "Night"]

employees = []

for i in range(1, NUM_EMPLOYEES + 1):

    dept = departments[i-1]

    employees.append({

        "employee_id": i,

        "employee_code": f"EMP{i:04d}",

        "first_name": fake.first_name(),

        "last_name": fake.last_name(),

        "gender": random.choice(["Male", "Female"]),

        "date_of_birth": fake.date_of_birth(minimum_age=21, maximum_age=58),

        "joining_date": fake.date_between(start_date="-10y", end_date="today"),

        "department": dept,

        "designation": random.choice(designations[dept]),

        "shift": random.choice(shifts),

        "plant": "SoluTex Manufacturing Plant",

        "manager_name": fake.name(),

        "salary": random.randint(180000,1200000),

        "phone": fake.phone_number(),

        "email": fake.company_email(),

        "employee_status": random.choice(employee_status)

    })

df = pd.DataFrame(employees)

print("Employee Table Generated Successfully!")
```

```python
print("Shape :", df.shape)

print("\nMissing Values")
print(df.isnull().sum())

print("\nDepartments")
print(df["department"].value_counts())

print("\nEmployee Status")
print(df["employee_status"].value_counts())

print("\nAverage Salary")
print(round(df["salary"].mean(),2))
```

```python
df.to_csv("dim_employee.csv", index=False)

df.to_excel("dim_employee.xlsx", index=False)

print("dim_employee.csv created successfully!")
print("dim_employee.xlsx created successfully!")
```
