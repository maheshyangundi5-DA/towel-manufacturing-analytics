```python
# Load Dimension Tables

products = pd.read_csv("dim_product.csv")

employees = pd.read_csv("dim_employee.csv")

machines = pd.read_csv("dim_machine.csv")

dates = pd.read_csv("dim_date.csv")

print("Dimension Tables Loaded Successfully!")
```

```python
quality_employees = employees[
    employees["department"] == "Quality"
].copy()

print("Quality Employees :", len(quality_employees))
```

```python
product_list = products.to_dict("records")

employee_list = quality_employees.to_dict("records")

machine_list = machines.to_dict("records")

date_list = dates.to_dict("records")

print("Lookup Lists Created Successfully!")
```

```python
inspection_result = {

    "Pass":90,

    "Rework":8,

    "Reject":2

}

print(inspection_result)
```

```python
defect_types = [

    "No Defect",

    "Color Variation",

    "Loose Thread",

    "Weaving Defect",

    "Incorrect Size",

    "Stain",

    "Hole",

    "Packaging Defect",

    "Label Error"

]

print(defect_types)
```

```python
product_lookup = {}

for product in product_list:

    product_lookup[product["product_id"]] = {

        "category": product["category"],

        "product_name": product["product_name"]

    }

print("Product Lookup Created Successfully!")
```

```python
def generate_quality(total_records):

    quality_data = []

    for quality_id in range(1, total_records + 1):

        # Progress
        if quality_id % 10000 == 0:
            print(f"Generated {quality_id:,} records...")

        # Random Dimension Records
        product = random.choice(product_list)

        employee = random.choice(employee_list)

        machine = random.choice(machine_list)

        date = random.choice(date_list)

        # Batch Number
        batch_number = "BATCH" + str(random.randint(100000,999999))

        # Inspection Quantities
        inspected_quantity = random.randint(100,5000)

        failed_quantity = random.randint(
            0,
            max(1,int(inspected_quantity*0.10))
        )

        passed_quantity = inspected_quantity - failed_quantity

        defect_percentage = round(
            (failed_quantity/inspected_quantity)*100,
            2
        )

        # Defect Type
        if failed_quantity == 0:
            defect_type = "No Defect"
        else:
            defect_type = random.choice(defect_types[1:])

        # Inspection Result
        result = random.choices(
            population=list(inspection_result.keys()),
            weights=list(inspection_result.values()),
            k=1
        )[0]

        # Inspection Time
        inspection_time = random.randint(5,60)

        # Quality Score
        quality_score = round(
            random.uniform(70,100),
            2
        )

        # Remarks
        if result=="Pass":
            remarks="Approved"

        elif result=="Rework":
            remarks="Send For Rework"

        else:
            remarks="Rejected"

        # Append Record
        quality_data.append({

            "quality_id":quality_id,

            "inspection_date_id":date["date_id"],

            "product_id":product["product_id"],

            "employee_id":employee["employee_id"],

            "machine_id":machine["machine_id"],

            "batch_number":batch_number,

            "inspected_quantity":inspected_quantity,

            "passed_quantity":passed_quantity,

            "failed_quantity":failed_quantity,

            "defect_percentage":defect_percentage,

            "defect_type":defect_type,

            "inspection_result":result,

            "inspection_time_minutes":inspection_time,

            "remarks":remarks,

            "quality_score":quality_score

        })

    df = pd.DataFrame(quality_data)

    return df
```
