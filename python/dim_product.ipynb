```python
import pandas as pd
from faker import Faker
import random

fake = Faker("en_IN")
```

```python
# =====================================================
# Manufacturing Production Intelligence Platform
# Table : dim_product
# =====================================================
```

```python
NUM_PRODUCTS = 1000

categories = (
    ["Bath Towel"] * 300 +
    ["Hand Towel"] * 200 +
    ["Face Towel"] * 150 +
    ["Bath Mat"] * 100 +
    ["Kitchen Towel"] * 100 +
    ["Hotel Towel"] * 100 +
    ["Beach Towel"] * 50
)

random.shuffle(categories)

colors = [
    "White","Blue","Grey","Red","Black",
    "Green","Brown","Pink","Orange","Purple"
]

sizes = [
    "30x30",
    "40x60",
    "50x90",
    "70x140",
    "75x150",
    "80x160"
]

gsm_list = [
    300,350,400,450,500,550,600
]

products = []

for i in range(1, NUM_PRODUCTS + 1):

    cost = random.randint(80,800)

    margin = random.uniform(1.20,1.80)

    selling_price = round(cost * margin,2)

    products.append({

        "product_id": i,

        "product_code": f"PRD{i:05d}",

        "product_name": categories[i-1] + " " + random.choice(colors),

        "category": categories[i-1],

        "size": random.choice(sizes),

        "gsm": random.choice(gsm_list),

        "color": random.choice(colors),

        "material": "100% Cotton",

        "supplier_id": random.randint(1,250),

        "unit_cost": cost,

        "selling_price": selling_price,

        "product_status": random.choice(["Active","Inactive"])

    })

df = pd.DataFrame(products)

print("Product Table Generated Successfully!")
```

```python
print("Shape :", df.shape)

print("\nMissing Values")
print(df.isnull().sum())

print("\nCategory Distribution")
print(df["category"].value_counts())

print("\nAverage Selling Price")
print(round(df["selling_price"].mean(),2))

print("\nSupplier Range")
print(df["supplier_id"].min(), "-", df["supplier_id"].max())
```

```python
df.to_csv("dim_product.csv", index=False)

df.to_excel("dim_product.xlsx", index=False)

print("dim_product.csv created successfully!")
print("dim_product.xlsx created successfully!")
```
