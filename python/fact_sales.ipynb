```python
import pandas as pd
import numpy as np
import random
from faker import Faker

fake = Faker("en_IN")
```

```python
# Load Dimension Tables

customers = pd.read_csv("dim_customer.csv")

products = pd.read_csv("dim_product.csv")

employees = pd.read_csv("dim_employee.csv")

dates = pd.read_csv("dim_date.csv")

print("Dimension Tables Loaded Successfully!")
```

```python
sales_employees = employees[
    employees["department"] == "Sales"
]

print("Sales Employees :", len(sales_employees))
```

```python
product_lookup = products.set_index("product_id")[
    [
        "selling_price",
        "unit_cost",
        "category"
    ]
].to_dict("index")
```

```python
quantity_rules = {
    "Retailer": (5,50),
    "Wholesaler": (100,500),
    "Hotel": (300,1200),
    "Exporter": (1000,5000),
    "Hospital": (50,300)
}

discount_rules = {
    "Retailer": (0,5),
    "Wholesaler": (5,10),
    "Hotel": (8,15),
    "Exporter": (10,20),
    "Hospital": (5,10)
}

status_weights = {
    "Completed":92,
    "Pending":5,
    "Cancelled":3
}
```

```python
sales_data = []

for sales_id in range(1,101):

    customer = customers.sample(1).iloc[0]

    customer_type = customer["customer_type"]

    qty = random.randint(
        quantity_rules[customer_type][0],
        quantity_rules[customer_type][1]
    )

    product = products.sample(1).iloc[0]

    price = product["selling_price"]

    cost = product["unit_cost"]

    discount_percent = random.randint(
        discount_rules[customer_type][0],
        discount_rules[customer_type][1]
    )

    gross_sales = qty * price

    discount_amount = round(
        gross_sales * discount_percent / 100,
        2
    )

    net_sales = gross_sales - discount_amount

    total_cost = qty * cost

    gross_profit = round(
        net_sales - total_cost,
        2
    )

    profit_margin = round(
        (gross_profit / net_sales) * 100,
        2
    )

    order_status = random.choices(
        population=list(status_weights.keys()),
        weights=list(status_weights.values())
    )[0]

    employee = sales_employees.sample(1).iloc[0]

    date = dates.sample(1).iloc[0]

    sales_data.append({

        "sales_id": sales_id,

        "order_number": f"ORD{100000 + sales_id}",

        "order_date_id": date["date_id"],

        "customer_id": customer["customer_id"],

        "product_id": product["product_id"],

        "sales_person_id": employee["employee_id"],

        "quantity": qty,

        "unit_price": price,

        "discount_percent": discount_percent,

        "discount_amount": discount_amount,

        "gross_sales": gross_sales,

        "net_sales": net_sales,

        "unit_cost": cost,

        "total_cost": total_cost,

        "gross_profit": gross_profit,

        "profit_margin": profit_margin,

        "order_status": order_status

    })

df = pd.DataFrame(sales_data)

print("Sales Data Generated Successfully!")
```
