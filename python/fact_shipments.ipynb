```python
def generate_shipments(total_records):

    shipment_data = []

    for shipment_id in range(1, total_records + 1):

        # Progress
        if shipment_id % 10000 == 0:
            print(f"Generated {shipment_id:,} records...")

        # Random Dimension Records
        customer = random.choice(customer_list)
        product = random.choice(product_list)
        warehouse = random.choice(warehouse_list)
        employee = random.choice(employee_list)
        date = random.choice(date_list)

        # Shipment Number
        shipment_number = "SHP" + str(random.randint(100000, 999999))

        # Quantity
        quantity = random.randint(1, 500)

        # Shipping Cost
        shipping_cost = round(random.uniform(50, 5000), 2)

        # Courier
        courier_name = random.choice(courier_list)

        # Dispatch Date
        dispatch_date = pd.to_datetime(date["full_date"])

        # Transit Days
        transit_days = random.randint(1, 10)

        # Delivery Date
        delivery_date = dispatch_date + pd.Timedelta(days=transit_days)

        # Delivery Status
        status = random.choices(
            population=list(delivery_status.keys()),
            weights=list(delivery_status.values()),
            k=1
        )[0]

        # Shipping Mode
        shipping_mode = random.choice(shipping_modes)

        # Vehicle Type
        vehicle_type = random.choice(vehicle_types)

        # Dispatch Remarks
        remarks = random.choice(dispatch_remarks)

        # Customer Rating
        customer_rating = random.randint(1, 5)

        shipment_data.append({

            "shipment_id": shipment_id,

            "shipment_date_id": date["date_id"],

            "customer_id": customer["customer_id"],

            "product_id": product["product_id"],

            "warehouse_id": warehouse["warehouse_id"],

            "employee_id": employee["employee_id"],

            "shipment_number": shipment_number,

            "quantity": quantity,

            "shipping_cost": shipping_cost,

            "courier_name": courier_name,

            "dispatch_date": dispatch_date.strftime("%Y-%m-%d"),

            "delivery_date": delivery_date.strftime("%Y-%m-%d"),

            "transit_days": transit_days,

            "delivery_status": status,

            "shipping_mode": shipping_mode,

            "vehicle_type": vehicle_type,

            "dispatch_remarks": remarks,

            "customer_rating": customer_rating

        })

    df = pd.DataFrame(shipment_data)

    return df
```
