```python
import pandas as pd
from faker import Faker
import random
```

```python
# ====================================================
# Manufacturing Production Intelligence Platform
# Table : dim_customer
# ====================================================
```

```python
fake = Faker("en_IN")

NUM_CUSTOMERS = 5000

customer_types = (
    ["Retailer"] * 2000 +
    ["Wholesaler"] * 1250 +
    ["Hotel"] * 750 +
    ["Exporter"] * 500 +
    ["Hospital"] * 500
)

random.shuffle(customer_types)

regions = ["North","South","East","West","Central"]

status = ["Active","Inactive"]

customer_data = []

for i in range(1, NUM_CUSTOMERS + 1):

    customer_data.append({

        "customer_id": i,

        "customer_code": f"CUST{i:05d}",

        "customer_name": fake.company(),

        "customer_type": customer_types[i-1],

        "industry": random.choice([
            "Retail",
            "Hospitality",
            "Healthcare",
            "Export",
            "Wholesale"
        ]),

        "city": fake.city(),

        "state": fake.state(),

        "country": "India",

        "region": random.choice(regions),

        "contact_person": fake.name(),

        "email": fake.company_email(),

        "phone": fake.phone_number(),

        "credit_limit": random.randint(50000,5000000),

        "customer_status": random.choice(status),

        "created_date": fake.date_between(
            start_date="-5y",
            end_date="today"
        )

    })

df = pd.DataFrame(customer_data)

print("Customer Table Generated Successfully!")
```

```python
df.head()
```

```python
print("Shape :", df.shape)

print("\nMissing Values")
print(df.isnull().sum())

print("\nCustomer Types")
print(df["customer_type"].value_counts())

print("\nAverage Credit Limit")
print(df["credit_limit"].mean())

print("\nCustomer Status")
print(df["customer_status"].value_counts())
```

```python
df.to_csv("dim_customer.csv", index=False)

df.to_excel("dim_customer.xlsx", index=False)

print("dim_customer.csv created successfully!")
print("dim_customer.xlsx created successfully!")
```
