```python
# ==========================================
# Generate dim_date Table
# ==========================================

# Date Range
start_date = "2023-04-01"
end_date = "2026-03-31"

# Generate all dates
dates = pd.date_range(start=start_date, end=end_date)

# Create DataFrame
df = pd.DataFrame()

# Primary Key
df["date_id"] = range(1, len(dates) + 1)

# Full Date
df["full_date"] = dates

# Day Details
df["day_number"] = dates.day
df["day_name"] = dates.day_name()

# Week Details
df["week_number"] = dates.isocalendar().week.astype(int)

# Month Details
df["month_number"] = dates.month
df["month_name"] = dates.month_name()

# Quarter
df["quarter_number"] = dates.quarter

# Calendar Year
df["year_number"] = dates.year

# Financial Year (India: April to March)
financial_year = []

for d in dates:
    if d.month >= 4:
        fy = f"FY{d.year}-{str(d.year + 1)[2:]}"
    else:
        fy = f"FY{d.year - 1}-{str(d.year)[2:]}"
    financial_year.append(fy)

df["financial_year"] = financial_year

# Weekend Flag
df["is_weekend"] = dates.weekday >= 5
df["is_weekend"] = df["is_weekend"].astype(int)

print("Data Generated Successfully!")
```

```python
# Additional Date Attributes

df["month_short"] = dates.strftime("%b")

df["quarter_name"] = "Q" + dates.quarter.astype(str)

df["day_of_week"] = dates.dayofweek + 1      # Monday=1

df["is_month_start"] = dates.is_month_start.astype(int)

df["is_month_end"] = dates.is_month_end.astype(int)

df["month_year"] = dates.strftime("%b-%Y")

df["year_month"] = dates.strftime("%Y-%m")
```
